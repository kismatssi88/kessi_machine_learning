# Tensors in ml:-

A tensor is a data container that can store numbers in multiple dimensions. In ML, tensors are used to represent things like datasets, images, text, audio, and model parameters.
| Type          | Dimensions | Example            | Common ML use                    |
| ------------- | ---------: | ------------------ | -------------------------------- |
| **Scalar**    |         0D | `5`                | Single value                     |
| **Vector**    |         1D | `[1, 2, 3]`        | Features of one sample           |
| **Matrix**    |         2D | `[[1,2], [3,4]]`   | Dataset / tabular data           |
| **3D Tensor** |         3D | `[[[1,2], [3,4]]]` | One RGB image / sequence data    |
| **4D Tensor** |         4D | Batch of images    | Multiple images in deep learning |
