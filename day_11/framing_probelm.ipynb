# Framing the Problem in Machine Learning

## 1. Business Problem

The first step in machine learning is to understand the business problem and identify what we want to achieve.

We should ask:
- What problem are we trying to solve?
- Why is this problem important?
- Who will use the predictions?
- What benefits will the solution provide?
- How will we measure success?

**Example:** An online shopping company wants to predict whether a customer will leave the platform.

## 2. Types of Problems

We need to identify the type of problem before selecting an ML algorithm.

### A. Classification
Predicts a category or class.

**Examples:**
- Spam or not spam
- Pass or fail
- Customer will leave or stay

### B. Regression
Predicts a continuous numerical value.

**Examples:**
- House price prediction
- Temperature prediction
- Sales forecasting

### C. Clustering
Groups similar data points without predefined labels.

**Examples:**
- Customer segmentation
- Grouping similar products

### D. Anomaly Detection
Identifies unusual patterns or observations.

**Examples:**
- Fraud detection
- Detecting unusual network activity

### E. Recommendation
Suggests relevant products, movies, or content to users.

**Examples:**
- Netflix movie recommendations
- E-commerce product recommendations

## 3. Current Solution

Before building an ML model, understand how the problem is currently being solved.

Ask:
- Is the problem solved manually?
- Is there an existing rule-based system?
- Is an existing ML model being used?
- What are the limitations of the current solution?
- Can machine learning improve the results?

**Example:** A bank currently detects fraudulent transactions using manually defined rules. An ML model could learn patterns from historical transactions and help identify suspicious activity.

**Important:** Machine learning is not always necessary. A simple rule-based solution may be cheaper, faster, and easier to maintain.

## 4. Getting the Data

After understanding the problem, identify and collect the required data.

### Sources of Data
- Databases
- CSV and Excel files
- APIs
- Websites, where permitted
- Sensors and devices
- User activity logs
- Public datasets, such as Kaggle datasets

### Important Considerations
- Is enough data available?
- Is the data relevant to the problem?
- Are there missing or incorrect values?
- Are the target labels available?
- Does the data represent real-world conditions?
- Do we have permission to use the data?

**Example:** To predict customer churn, we might collect customer tenure, purchase history, complaints, subscription details, and whether the customer eventually left.

## 5. Metrics to Measure Success

Metrics help us evaluate whether the model solves the problem effectively.

### A. Classification Metrics

- **Accuracy:** Percentage of all predictions that are correct.
- **Precision:** Of all predicted positive cases, how many are actually positive?
- **Recall:** Of all actual positive cases, how many did the model identify?
- **F1-score:** Harmonic mean of precision and recall.

### B. Regression Metrics

- **MAE (Mean Absolute Error):** Average absolute difference between actual and predicted values.
- **MSE (Mean Squared Error):** Average squared prediction error.
- **RMSE (Root Mean Squared Error):** Square root of MSE.

### C. Business Metrics

- Revenue increase
- Cost reduction
- Customer retention rate
- Fraud losses prevented
- Time saved

**Important:** Choose metrics based on the problem. For example, in fraud detection, recall and precision may be more useful than accuracy alone.

## 6. Online Learning vs Batch Learning

This describes how an ML model learns or receives updates.

### A. Batch Learning

The model is trained using a dataset in a training process. When new data arrives, the model generally needs to be retrained periodically to learn from it.

**Advantages:**
- Easier to manage and evaluate.
- Suitable when data changes slowly.
- Training can use large amounts of historical data.

**Disadvantages:**
- Retraining can be expensive.
- The model may become outdated between training runs.

**Examples:**
- House price prediction models retrained monthly.
- Product demand models retrained weekly.

### B. Online Learning

The model learns incrementally from new data, updating itself continuously or in small batches.

**Advantages:**
- Adapts to changing patterns.
- Can process continuously arriving data.
- May require less memory than training on the entire dataset at once.

**Disadvantages:**
- Requires careful monitoring.
- Poor-quality data can affect future learning.
- Changes in data patterns can cause problems if not managed properly.

**Examples:**
- Updating a model as new transaction patterns arrive.
- Learning from continuously arriving sensor data.

**Key difference:** Batch learning updates the model through periodic training, while online learning updates it incrementally as new data arrives.

*Note:* In production systems, online versus batch can also refer to how predictions are served: real-time predictions versus scheduled predictions. This is a separate distinction from how a model learns.

## 7. Check Assumptions

Before building the model, verify the assumptions made about the problem and data.

### Important Checks

1. **Data availability:** Is enough relevant data available?
2. **Data quality:** Are missing values, duplicates, and incorrect values handled?
3. **Representativeness:** Does the dataset reflect the real-world population?
4. **Feature availability:** Will the input features be available when making predictions?
5. **Data leakage:** Does any feature reveal information that would not be available at prediction time?
6. **Target definition:** Is the target variable clearly defined?
7. **Relationship stability:** Are the patterns likely to remain useful over time?
8. **Evaluation method:** Does the train-test split match the intended real-world use?
9. **Ethical and legal constraints:** Is the data collected and used appropriately?

**Example:** If we predict whether a student will pass an exam, using the final exam marks as an input feature would cause data leakage because those marks are not available before the exam result.

## Summary

Framing an ML problem involves:

1. Defining the business problem.
2. Identifying the type of ML problem.
3. Understanding the current solution.
4. Collecting the required data.
5. Selecting suitable evaluation metrics.
6. Deciding between batch and online learning.
7. Checking assumptions before developing the model.

**Remember:** A successful ML project starts with the right problem, suitable data, meaningful metrics, and realistic assumptions—not simply choosing an algorithm.
